## **Question 3:** (5 Points)

Replace any missing values in the `experience` column with the median of that column.

```python
median_experience = employees_df["experience"].median()
employees_df["experience"] = employees_df["experience"].fillna(median_experience)

print(employees_df)
```

## **Question 4:** (5 Points)

Convert the `salary` column to a numeric data type, handling any non-numeric values by replacing them with the mean salary.

```python
# Turn salary into numbers.
employees_df["salary"] = pd.to_numeric(employees_df["salary"], errors='coerce')

mean_salary = employees_df["salary"].mean()

# Fill the odd spots.
employees_df["salary"] = employees_df["salary"].fillna(mean_salary)
print(employees_df)
```

## **Question 5:** (5 Points)

Rename the column `education level` to `Education_level`.

```python
# New column name.
employees_df = employees_df.rename(columns={"education level": "Education_level"})

print(employees_df.columns)
```

## **Question 6:** (5 Points)

Find the average age of employees in the dataframe.

```python
# Average age.
average_age = employees_df["age"].mean()

print(average_age)
```

## **Question 7:** (10 Points)

Filter the dataframe to display only those employees who have a Master's degree or higher in the `education_level` column.

```python
# Master's or better.
higher_ed = employees_df[
    employees_df["Education_level"].isin(["Master", "PhD"])
]

print(higher_ed)
```

## **Question 8:** (10 Points)

Using the `apply` function, create a new column `experience_category` that categorizes employees as "Junior" if they have less than 3 years of experience, "Mid-level" if they have 3 to 7 years, and "Senior" if they have more than 7 years.

```python
# function for the labels.
def experience_label(x):
    if x < 3:
        return "Junior"
    elif x <= 7:
        return "Mid-level"
    else:
        return "Senior"

# New column.
employees_df["experience_category"] = employees_df["experience"].apply(experience_label)

print(employees_df)
```

## **Question 9:** (10 Points)

Identify and remove any duplicate rows from the dataframe.

```python
# Drop duplicate rows.
employees_df = employees_df.drop_duplicates()

print(employees_df)
```

## **Question 10:** (10 Points)

Find employees who have "PhD" in their education but a salary less than the mean salary of the dataframe.

```python
# Mean salary first.
mean_salary = employees_df["salary"].mean()

# PhD but making less than average.
low_paid_phd = employees_df[
    (employees_df["Education_level"] == "PhD") &
    (employees_df["salary"] < mean_salary)
]

print(low_paid_phd)
```

## **Question 11:** (10 Points)

Sort the dataframe based on the `experience` column in descending order.

## **Question 12:** (10 Points)

Group the dataframe by `education_level` and find the average salary for each education level.

```python
# Average salary by education.
avg_salary_by_education = employees_df.groupby("Education_level")["salary"].mean()

print(avg_salary_by_education)
```

## **Question 13:** (10 Points)

Write a function that categorizes each employee's experience into 'Novice', 'Intermediate', or 'Expert' based on the number of years of experience:

- 'Novice' for less than 5 years.
- 'Intermediate' for 5 to 10 years.
- 'Expert' for more than 10 years.

Consider missing experience values as 'Unknown' and handle them appropriately in your function.

Apply this function to the experience column to create a new column named experience_level.

```python
# Experience labels again, but different.
def experience_type(x):
    if x < 5:
        return "Novice"
    elif x <= 10:
        return "Intermediate"
    else:
        return "Expert"

employees_df["experience_level"] = employees_df["experience"].apply(experience_type)

print(employees_df)
```
