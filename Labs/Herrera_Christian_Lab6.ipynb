## Game Character and Hero Hierarchy

### Objective:
You will create a superclass named `GameCharacter` and a subclass named `Hero`, which extends the functionality of `GameCharacter`. You'll use conditional statements and loops to manage the character’s actions and interactions with other characters in a game-like scenario.

#### Tasks:

1. **GameCharacter Class (15 Points)**:
   - Create a superclass named `GameCharacter` with the following attributes:
       - `name` (string)
       - `health` (integer)
       - `attack_power` (integer)
   - Implement an `__init__` method that initializes a `GameCharacter` with `name`, `health`, and `attack_power`.
   - Include the following methods in the `GameCharacter` class:
       - `display_info()`: Prints the character's `name`, `health`, and `attack_power`.
       - `take_damage(damage)`: Reduces the character’s `health` by the `damage` amount and prints the character’s updated `health`. If the `health` falls to or below 0, print that the character has fainted.
       - `attack(other_character)`: Causes this character to attack another character. This should reduce the other character’s health by this character’s `attack_power` and print the action.
   
2. **Hero Class (15 Points)**:
   - Create a subclass named `Hero` that inherits from `GameCharacter`. Add the following additional attributes:
       - `special_power` (string)
       - `special_attack_power` (integer)
       - `mana` (integer)
   - Override the `__init__` method to initialize the new attributes as well as the inherited ones.
   - Override the `display_info()` method to include all the attributes from the `Hero` class.
   - Implement a `use_special_attack(other_character)` method that allows the `Hero` to use their special power to deal `special_attack_power` damage to `other_character`. However, this can only be done if the `Hero` has at least 10 mana points (costs 10 mana to use the special attack). If they don’t have enough mana, print a message that they cannot use their special power.
   - Implement a `mana_regen(amount)` method that regenerates the `Hero's` mana points by the specified `amount` and prints the updated mana level.
   
3. **Battle Simulation (15 Points)**:
   - Implement a `battle(other_character)` method for the `Hero`. In this method:
       - The `Hero` will repeatedly attack the `other_character` using either regular attacks or the special attack (if enough mana) until one of the characters’ health reaches zero.
       - Use a **loop** to perform the turn-based attack sequence, where both characters take turns attacking each other. The `Hero` will always attack first.
       - On each turn, print out the current health of both characters.
       - The loop should stop when one character’s health is zero or less, and the method should declare the winner.

"4. **Testing Your Code (15 Points)**:
   - Create a `Hero` and a `GameCharacter` instance. Demonstrate the use of all methods including:
     - Taking and dealing damage.
     - Using the `Hero`'s special attack when enough mana is available.
     - Simulating a battle between the two characters.
"
For Example:
- hero1 = Hero("Knight", 120, 25, "Sword Slash", 50, 30)
- villain1 = GameCharacter("Goblin", 80, 15)

Run the `battle()` method to simulate a battle between the hero and the villain. Observe how the game dynamics, conditionals, and loops manage the flow of the battle.


```python
class GameCharacter:
    def __init__(self, name: str, health: int, attack_power: int): # Initializing with the attribbutes. 
        self.name = name
        self.health = health
        self.attack_power = attack_power
            
    def display_info(self):
        print(f'Name: {self.name}, HP: {self.health}, Normal attack power:{self.attack_power}')    # Simple display function. 
    def take_damage(self, damage): # Taking damage, and fainting when we hit zero. 
        self.health -= damage
        print(f"{self.name}'s health is now {self.health}")
        if self.health <= 0:
            print(f"{self.name} has fainted.")  
    def attack(self, other_character): # attacking other people, unprovoked. 
        other_character.take_damage(self.attack_power)
        print(f"{self.name} inflicts {self.attack_power} points of damage on {other_character.name}.")
                                                    
                        
class Hero(GameCharacter): # The Hero (me)
    def __init__(self, name, health, attack_power, special_power, special_attack_power, mana):
        super().__init__(name, health, attack_power)
        
        self.special_power = special_power
        self.special_attack_power = special_attack_power
        self.mana = mana
    def display_info(self): # Adding on to the original display function. 
        super().display_info()
        print(f'Special attack: {self.special_power}, Special attack power: {self.special_attack_power}, Mana: {self.mana}')
        
    def use_special_attack(self, other_character):
        # costs 10 mana
        if self.mana < 10:
            print(f"{self.name} cannot use {self.special_power} (not enough mana).")
            return
        self.mana -= 10
        other_character.take_damage(self.special_attack_power)
        print(
            f"{self.name} uses {self.special_power} and inflicts {self.special_attack_power}  points of damage to {other_character.name}!"
        )
        print(f"{self.name}'s mana is now {self.mana}")
        
    def mana_regen(self, amount): # Regenerating mana
        self.mana += amount
        print(f"{self.name}'s mana is now {self.mana}")
        
    def battle(self, other_character): # The battle.
        # Hero always attacks first.
        while self.health > 0 and other_character.health > 0:
            print("\n Turn Start")
            print(f"{self.name} health: {self.health} | mana: {self.mana}")
            print(f"{other_character.name} health: {other_character.health}")
            # Hero turn: special if enough mana, otherwise regular attack
            if self.mana >= 10:
                self.use_special_attack(other_character)
            else:
                self.attack(other_character)
            if other_character.health <= 0:
                print(f"\nWinner: {self.name}")
                return
            # Other character turn
            other_character.attack(self)
            if self.health <= 0:
                print(f"\nWinner: {other_character.name}")
                return
# Create instances
hero1 = Hero("Knight", 120, 25, "Party Rock", 50, 30)
bad_guy1 = GameCharacter("Goblin", 80, 15)

hero1.display_info()
bad_guy1.display_info()

hero1.attack(bad_guy1)
bad_guy1.attack(hero1)

hero1.take_damage(10)

hero1.mana_regen(5)

hero1.use_special_attack(bad_guy1)

# another test
battle_hero = Hero("Solid Snake", 120, 25, "Sneak Attack", 50, 30)
battle_villain = GameCharacter("Liquid Snake", 80, 15)
battle_hero.battle(battle_villain)
```

## **Dice Battle Game**
Youhave to create a simple turn-based **Dice Battle Game** where two players (user inputs) compete against each other. The game will use user input for choices, conditionals to control the game flow, and loops to manage the rounds until a player wins.

#### Game Description:

- Both players will start with an initial health of 100.
- Each round, players will take turns rolling a dice (random number between 1 and 6).
- The dice roll result will determine how much damage the other player takes.
- Players will have the option to "heal" (regain some health) instead of attacking, but they can only heal once every 3 turns.
- The game continues until one player's health drops to 0 or below.

### Tasks:

1. **Player Class (15 Points)**:
   - Create a class named `Player` with the following attributes:
     - `name` (string): The player's name.
     - `health` (integer): The player's starting health, initialized to 100.
     - `can_heal` (boolean): A flag to track if the player is allowed to heal (they can heal only once every 3 turns).
     - `heal_counter` (integer): Tracks the number of turns since the player last healed.
   - Implement the following methods:
     - `roll_dice()`: Simulates a dice roll and returns a random number between 1 and 6.
     - `attack(other_player)`: Rolls the dice and deals the dice value as damage to the other player.
     - `heal()`: Allows the player to heal by a random value between 10 and 30 health points. The player can only heal once every 3 turns. If they can heal, reset `can_heal` to `False` and `heal_counter` to 0.
     - `end_turn()`: Increments the `heal_counter`. If it reaches 3, set `can_heal` to `True` so the player can heal again.

2. **Game Loop (15 Points)**:
   - Create a loop to alternate between two players’ turns.
   - On each turn:
     - Ask the current player whether they want to **attack** or **heal**.
     - Perform the selected action (attack the other player or heal).
     - Print the current health of both players after each turn.
     - If a player's health reaches 0 or below, the game ends and declares the other player the winner.

3. **User Input (10 Points)**:
   - Use input statements to take the players' names at the beginning.
   - Use input on each turn to ask whether the player wants to attack or heal.



```python
import random

class Player:
    def __init__(self, name, health=100, can_heal=True, heal_counter=3): # Initializing
        self.name = name
        self.health = health
        self.can_heal = can_heal
        self.heal_counter = heal_counter

    def roll_dice(self): 
        return random.randrange(1, 7, 1) # My preferred random number generator. 
    def take_damage(self, damage): # Taking damage. 
        self.health -= damage
        print(f"{self.name}'s health is now {self.health}")
        if self.health <= 0:
            print(f"{self.name} has fainted.")
    def attack(self, other_player): # Dealing damage
        damage = self.roll_dice()
        other_player.take_damage(damage)
        print(f"{self.name} rolls {damage} ad attacks {other_player.name}!")

    def heal(self):
        if self.can_heal:
            amount = random.randrange(10, 31, 1)
            self.health += amount
            print(f"{self.name} heals for {amount}! Health is now {self.health}")
            self.can_heal = False
            self.heal_counter = 0
        else:
            print(f"{self.name} cannot heal right now!")
    def end_turn(self): # The end.
        self.heal_counter += 1
        if self.heal_counter >= 3:
            self.can_heal = True

# Get names and stuff
p1_name = input("Player 1 name: ")
p2_name = input("Player 2 name: ")

# Make the players
player_1 = Player(p1_name, 100, True, 3)
player_2 = Player(p2_name, 100, True, 3)

current_player = player_1
other_player = player_2

# Main game loop
while player_1.health > 0 and player_2.health > 0:
    
    # Whose turn?
    print(f"{current_player.name}'s turn.")
    print(f"{player_1.name} HP: {player_1.health} | {player_2.name} HP: {player_2.health}")

    choice = input("Attack or heal? (a/h): ").lower()

    # do the thing
    if choice == "h":
        current_player.heal()
    else:
        # Default to attack if they type weird stuff!!
        current_player.attack(other_player)

    # End turn heal timer stuff
    current_player.end_turn()

    # Show hp after turn
    print(f"After turn -> {player_1.name} HP: {player_1.health} | {player_2.name} HP: {player_2.health}")

    # check win?
    if other_player.health <= 0:
        print(f"\n{current_player.name} wins!")
        break

    # swap turns
    current_player, other_player = other_player, current_player
```
