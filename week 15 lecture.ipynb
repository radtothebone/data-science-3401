```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Lectures/customers.csv')
print('Dataframe shape:', df.shape)
df.head()
```

```python
class DataProcessor:
    def __init__(self, data):
        self.data = data
    
    def preprocess_data(self):
        # Handling missing values.
        self.data.dropna(inplace=True)
        
        # Making sure age values are integers. 
        self.data['Age'] = self.data['Age'].astype(int)

    def display_data(self):
        print(self.data)
```

```python
# Create an instance of our DataProcessor class.
processor = DataProcessor(df)

processor.preprocess_data()

processor.display_data()
```

```python
# Example 2: Custom data transformation class

class DataTransformer:
    def __init__(self, data):
        self.data = data
    
    def transform_age(self):
        def categorize_age(age):
            if age < 30:
                return 'Young'
            elif 30 <= age <= 60:
                return 'Middle-aged'
            else:
                return 'Old'
        
        self.data['Age_Group'] = self.data['Age'].apply(categorize_age) # Very important line of code!
    
```

```python
transfromer = DataTransformer(df)

transfromer.transform_age()

print(transfromer.data)
```

```python
# Data analysis class

class DataAnalysis:
    def __init__(self, data):
        self.data = data
    
    def summary_statistics(self):
        summary = {
        'mean_age': self.data['Age'].mean(),
        'median_age': self.data['Age'].median(),
        'min_age': self.data['Age'].min(),
        'max_age': self.data['Age'].max()
        }

        return summary
    
test = DataAnalysis(df)

print(test.summary_statistics())
```

```python
# Example 4: Advanced data set analyzer. 

class AdvDataAnalyzer:
    def __init__(self, data:pd.DataFrame):
        self.data = data
    
    def dataset_dimension(self):
        return self.data.shape
    
    def column_names(self):
        return self.data.columns.to_list()
    
    def dataframe_head(self, rows=10):
        return self.data.head(rows)

    def dataframe_tail(self, rows=10):
        return self.data.tail(rows)
    
class SpecialAnalyzer(AdvDataAnalyzer):
    def check_missing_values(self):
        return self.data.isnull().sum()
    
    def check_duplicate_values(self):
        return self.data.duplicated().sum()
    
    def remove_duplicates(self):
        return self.data.drop_duplicates()
    
    def fill_missing(self, value=0):
        return self.data.fillna(value) # How to do with with the mean value?
    
    def summary_statistics(self):
        return self.data.describe()

advanced_data_analyzer = AdvDataAnalyzer(df)

advanced_data_analyzer.dataframe_head()
```
