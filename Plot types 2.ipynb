```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import plotly.express as px

n = np.array([0,1,2,3,4,5])


xx=np.arange(-1,1,0.1)

fig,axes = plt.subplots(1,4, figsize=(12,4))

axes[0].scatter(xx,xx + 0.1*np.random.randn(len(xx)))
axes[0].set_title('Scatter plot')

axes[1].step(n, n**2, lw=4)
axes[1].set_title('Step plot')

axes[2].bar(n, n**2, align = 'center', width = 0.5)
axes[2].set_title('Bar Plot')

axes[3].fill_between(xx, xx**2, xx**3, color='green')
axes[3].set_title('Fill Between')
```

```python
# Seaborn!!
import seaborn as sns

# Practice data sets
tips = sns.load_dataset('tips')

# Bar plot

sns.barplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
sns.boxplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
#Kernel density estimation
sns.kdeplot(tips['total_bill'])
plt.show()
```

```python
sns.histplot(tips['total_bill'], kde=True, bins=14)
plt.show()
```

```python
sns.violinplot(x='day', y='total_bill', data=tips)
plt.show()
```

```python
data = pd.read_csv('Movies-2.csv')
data.head()
```

```python
# annotating graphs with arrows and lines and such
x_coord = 0
y_coord = 0

text_x = 40
text_y = 10

# Create the scatter plot
plt.scatter(data['RottenTomatoes_Ratings'], data['Audience Ratings'])

# Adding the annotation
plt.annotate('Annotated point', 
             xy=(x_coord,y_coord), 
             xytext=(text_x,text_y), 
             arrowprops=dict(facecolor='black', shrink=0.03))

plt.show()
```

```python
# The heatmap
correlation_matrix = data.corr(numeric_only=True)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=5)
```

```python
# Creating pair plots

columns_to_plot = ['RottenTomatoes_Ratings', 'Audience Ratings', 'Budget_m', 'Genre']
sns.pairplot(data[columns_to_plot], markers=['o', 's', 'D'], height=5)
plt.suptitle('Pair Plots', size=20)
plt.tight_layout()
plt.show()
```

```python
data_mpg = sns.load_dataset('mpg')
g = sns.PairGrid(data_mpg)
g.map_lower(sns.regplot)
g.map_upper(sns.scatterplot)
```
