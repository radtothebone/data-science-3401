```python
a=np.array([1,2,3,4,5])
sum_a=np.sum(a)
print(sum_a)
```

```python
array_mean=np.mean(a)
print(array_mean)
array_max=np.max(a)
print(array_max)
```

```python
i=np.array([1,2,3])
j=np.array([4,5,6])

#Broadcasting:adding a scalar to an array
k=i+2
print(k)

l=i*j
print(l)
```
