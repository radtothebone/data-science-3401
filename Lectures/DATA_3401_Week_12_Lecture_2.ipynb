```python
# sub data set

sub_covid_df=pd.read_csv('covid.csv',usecols=['date','state','death','positive','negative','totalTestResults'],parse_dates=['date'],infer_datetime_format=True)
sub_covid_df.head()
```

```python
# Problem with Null values
#isnull

pd.isnull(sub_covid_df)
```

```python
# To select just the rows with NaN values, we can use "any()" method

sub_covid_df[pd.isnull(sub_covid_df).any(axis=1)]
```

```python
# checking for missing values
print(sub_covid_df.isnull().sum())
```

```python
#Calculate the percentage of missing values for each column
missing_percentage=sub_covid_df.isnull().mean()*100
print(missing_percentage)
```

```python
#Drop rows with any missing values
sub_covid_df.dropna(inplace=True)
sub_covid_df
```

```python
# Drop rows with missing values in specific columns e.g: death and positive
sub_covid_df.dropna(subset=['death','positive'],inplace=True)
sub_covid_df
```

```python
# Fill missing values with 0
sub_covid_df.fillna(0,inplace=True)
sub_covid_df
```

```python
#Fill Missing values with the Mean, Median, Mode
sub_covid_df['death']=sub_covid_df['death'].fillna(sub_covid_df['death'].mean())
sub_covid_df['positive']=sub_covid_df['positive'].fillna(sub_covid_df['positive'].median())
sub_covid_df['state']=sub_covid_df['state'].fillna(sub_covid_df['state'].mode()[0])
```

```python
#checking for duplicates
print(sub_covid_df.duplicated().sum())
```

```python
# Drop duplicated rows, if any:
sub_covid_df.drop_duplicates(inplace=True)
```

```python
# Write DateFrame to outfile

sub_covid_df.to_csv('sub_covid_df.csv')
```

```python
sub_covid_df=sub_covid_df.sort_values(by='date')
sub_covid_df
```
