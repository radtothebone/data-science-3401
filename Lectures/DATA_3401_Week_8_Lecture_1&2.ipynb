## A Better Way to Open a File

Using the <code>with</code> statement is better practice, it automatically closes the file even if the code encounters an exception. The code will run everything in the indent block then close the file object.

```python
# Open file by using the "with" statement
with open('data.txt','r') as file:
  content=file.read()
  print(content)
```

# **Exception Handling**

### **Objectives**

After completing this lab you will be able to:

* Understand exceptions    
* Handle the exceptions

Python uses special objects called exceptions to manage errors that arise during a program’s execution. Whenever an error occurs that makes Python unsure of what to do next, it creates an exception object. If you write code that handles the exception, the program will continue running. If you don’t handle the exception, the program will halt and show a traceback, which includes a report of the exception that was raised.

- Exceptions are handled with **try-except** blocks.

- A try-except block asks Python to do something, but it also tells Python what to do if an exception is raised. When you use try-except blocks, your programs will continue running even if things start to go wrong.

## **What is an Exception?**

An exception is an error that occurs during the execution of code. This error causes the code to raise an exception and if not prepared to handle it will halt the execution of the code.

# Additional modes

1. r+: Reading and Writing. Cannot truncates the file.
2. w+: Writing and Reading. Truncates the file.
3. a+: Appending and Reading. Creates a new file, if none exists.

```python
# Open a file in ('r+') read/write mode
with open('data.txt','r+') as file:
  content= file.read() # Read the existing content
  print("Before writing:", content)

  file.write("\nAdding a new line while reading the file.")

  #Go back to the beginning and read the update file
  file.seek(0)
  updated_content=file.read()
  print("After writing:", updated_content)
```

```python
with open('data3.txt','a+') as file:
  file.write("This is another line E\n")
  file.seek(0)# #Move the file pointer to the beginning.
  print(file.read())
```

```python
with open('data3.txt','a+') as f:
  print("Initial Location:{}".format(f.tell()))
  data=f.read()# Read the file content from the current position(which is at the end in "a+" mode)
  if not data: # if data is empty
    print("Read noting or the pointer is likely at the end of the file")
  else:
    print(data)

  f.seek(0)# seek to the start of the file.

  print("\nNew Location {}".format(f.tell()))

  data=f.read() #reading from the start of the file after moving the pointer

  if not data: # check if the file is empty or no content was read
    print("Read nothing (the file might still be empty.)")
  else:
    print(data)

  print("Location after read:{}".format(f.tell()))# this should now be at the end of the file after reading all content
```

```python
def divide_numbers(a,b):
   try:
       result=a/b
       print(f"The result of {a} divided by {b} is {result}")
   except Exception as e: # block catches any exception that might occur during the division opration.
       print("An error occurred:", e)
```

```python
divide_numbers(10,2)
divide_numbers(5,0)
divide_numbers(10,"10")
```

```python
import datetime
def calculate_age(birth_year):
    current_year=datetime.datetime.now().year
    return current_year-birth_year

def process_ages(input_file, output_file):
  try:
     with open(input_file,'r') as file:
       lines=file.readlines()

     with open(output_file,'w') as file:
       for line in lines:
          if "," in line:
             try:
                 name, year= line.strip().split(",")
                 age=calculate_age(int(year))
                 file.write(f"{name} is {age} year old\n")
             except ValueError:

                 file.write(f"Error:Invalied year for {name}.\n")
          else:
              if line.strip():
                 file.write(f"Error:Line format incorrect and skipped:{line.strip()}\n")
  except FileNotFoundError:
      print(f"Error:Input file '{input_file}' not found.")
  except IOError:
      print(f"Error: An error occurred while reading or writing files")
  except Exception as e:
      print(f"An unexpected error occurred: {e}")
```
