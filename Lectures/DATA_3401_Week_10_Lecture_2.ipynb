```python
import numpy as np
import matplotlib.pyplot as plt

size=100000
x=np.random.randn(size)
plt.hist(x, bins=100, range=(-5,5))
plt.show() # displays the generated histogram plot.
```

## Beta Distribution

```python
x=[]
for i in range(10000):
   x.append(rnd.betavariate(4,4))

plt.hist(x,bins=50, range=(0,1))
plt.show()
```

## Uniform Distribution

```python
a= 5
b=20
x=np.random.uniform(0,1,size=100000)
y=(b-a)*x+a
plt.hist(y,bins=100,range=(a,b))
plt.show()
```

```python
#Concatenation
a=np.array([1,2,3])
b=np.array([4,5,6])
c=np.concatenate((a,b))
print(c)
```

```python
#Splitting
d=np.array([4,5,6,7,8,9,11,12,13])
e,f,g=np.split(d,3)
print(e)
print(f)
print(g)
```

```python
x=np.linspace(0,10,100)
y=np.sin(x)
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("Sin(x)")
plt.title("Sine function wave")
plt.show()
```

```python
# Matrix Inversion
a=np.array([[1,2],[3,4]])
inverse_a=np.linalg.inv(a)
print(inverse_a)
```

```python
# Matrix Determinant
detern_a=np.linalg.det(a)
print(detern_a)
```
