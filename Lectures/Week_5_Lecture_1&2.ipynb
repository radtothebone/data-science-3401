# Week 5-Data Structure

* 1. Primitive Data Structures

 int: integer values (eg. 1,2,100)

 float:floating-point values (1.0,2.0, 3.14)

 bool:boolean values(True or False)

 str: strings which are sequences of characters (eg."Hello, Python")

 * 2. Non-Primitive (Compound) Data Structures:

 1. **List:** They are defined using square brackets []
 2. **Tuple:** Its elements cannot be modified. They are defined using parentheses ()
 3. **Dictionary:** A collection of key-value pairs.They are defined using braces{}
 4. **Set**{} or set()


- **List:** A dynamic, ordered collection of elements. It allows storing elements of different types and is mutable (can be changed).They are defined using square brackets [].

- **Tuple:**  An ordered, immutable collection of elements. Once a tuple is created, its elements cannot be modified.They are defined using parentheses ().

- **Dictionary:**  A collection of key-value pairs. It is unordered, mutable, and allows for fast lookups by key.Dictionaries are defined using curly braces {} with keys and values.

- **Set:** An unordered collection of unique elements. It is mutable, but it does not allow duplicate values. Sets are defined using curly braces {} or the set() function, but unlike dictionaries, they only contain keys, not key-value pairs.

**list**
 1. order matters;
 2. need to change the data value
 3. may add or remove elements

 **Tuple**
 1. Date should not change.
 2. want faster performance than lists.
 3. The data represents a fixed structure.

 **Dictionary**
 1. fast lookups
 2. data is naturally in key-value form.

**Set**
1. need unique values.
2. order does not matter.
3. want fast membership tests.




| Feature           | List         | Tuple      | Dictionary                                                             | Set              |
| ----------------- | ------------ | ---------- | ---------------------------------------------------------------------- | ---------------- |
| Ordered           | Yes          | Yes        | No (Python 3.7+ preserves insertion order, but conceptually unordered) | No               |
| Mutable           | Yes          | No         | Yes                                                                    | Yes              |
| Allows duplicates | Yes          | Yes        | Keys: No                                                               | No               |
| Access method     | Index        | Index      | Key                                                                    | No index         |
| Syntax            | `[ ]`        | `( )`      | `{key: value}`                                                         | `{ }` or `set()` |
| Main use          | Ordered data | Fixed data | Key-value lookup                                                       | Unique elements  |

List with loops and conditional statements

```python
# Give me the sum of all positive numbers in the list
def sum_of_positive(numbers):
  total=0
  for num in numbers:
    if num >0:
      total+=num
  return total
```

```python
numbers=[1,-4,7,12,-9,0,5,-6]
print(sum_of_positive(numbers))
```

```python
# create a function that takes a list of numbers and returns a new list containing only the positive number.
def filter_of_positive(numbers):
  positive_num=[]
  for num in numbers:
    if num >0:
      positive_num.append(num)
  return positive_num
```

```python
numbers=[1,-4,7,12,-9,0,5,-6]
print(filter_of_positive(numbers))
```

```python
# Write a function that squares all even numbers in the list and keeps the odd number unchanged.
def square_even_num(numbers):
  result=[]
  for num in numbers:
    if num %2 ==0: # this is an even number
      result.append(num **2)
    else:  # other odd numbers
      result.append(num)
  return result
numbers=[1,-4,7,12,-9,0,5,-6]
print(square_even_num(numbers))
```

```python
# Finding the maximum value in a Tuple
def find_max(values):
   max_value=values[0]
   for i in values:
     if i>max_value:
        max_value=i
   return max_value
```

```python
values=(2,25,3,12,56,14,20)
print(find_max(values))
```

```python
# A Dictionary in a dictionary

user={
    'aeinstein':{
        'first':'albert',
        'last':'einstein',
        'location':'princeton'
    },

    'mcurie':{
        'first':'marie',
        'last':'curie',
        'location':'paris'
    }
    }

for username, user_info in user.items():
  print(username)
  full_name=f"{user_info['first']} {user_info["last"]}"
  location=user_info['location']
  print(full_name.title())
  print(location.title())
```
