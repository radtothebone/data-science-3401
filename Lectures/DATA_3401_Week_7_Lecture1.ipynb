# Week 7- Advance Object Oriented Programming (Advance OOP)



```python
# Super class: Pizza class is a base class that models a generic pizza
class Pizza:
 def __init__(self,name):
   self.name=name
   self.topping=[]

 def add_topping(self, topping):
   self.topping.append(topping)

 def bake(self):
   print(f"Baking {self.name} Pizza with topping {','.join(self.topping)}...")
```

```python
# Defines a subclass of Pizza
class MargheritaPizza(Pizza):
  def __init__(self):
    super().__init__("Margherita")
    self.add_topping("Tomato")
    self.add_topping("Mozzarella")
```

```python
#Defines another subclass of Pizza
class PepperoniPizza(Pizza):
  def __init__(self):
    super().__init__("Pepperoni")
    self.add_topping("Red Sauce")
    self.add_topping("Cheese")
    self.add_topping("Pepperoni")

```

```python
margherita=MargheritaPizza()
margherita.bake()
```

```python
Pepperoni=PepperoniPizza()
Pepperoni.bake()
```

# Example 1. Buliding on the vehicle system

```python
class Vehicle:
  def __init__(self, make,year,mileage):
    self.make=make
    self.year=year
    self.mileage=mileage

  def display_info(self):
    print(f"Vehicle Info:{self.year} {self.make}, Mileage:{self.mileage} miles.")

  def start_engine(self):
    print("Enginie started")

  def set_mileage(self, mileage):
    if mileage >= self.mileage:
      self.mileage=mileage
      print(f"New mileage set to {self.mileage} miles.")
    else:
      print("Error: New mileage cannot be less than the current mileage")

  def get_mileage(self):
    return self.mileage

  def drive(self, distance):
    if distance >0:
      self.mileage += distance
      print(f"Driving {distance} miles. New mileage is {self.mileage} miles.")
    else:
      print("Distance must be greater than zero")

#Car subclass definition

class Car (Vehicle):
  def __init__(self, make, year, mileage, doors, fuel_type):
    super().__init__(make,year,mileage)
    self.doors=doors
    self.fuel_type=fuel_type

  def display_info(self):
    super().display_info()
    print(f"Doors:{self.doors}, Fuel Type: {self.fuel_type}")

  def lock_doors(self):
    print("Doors locked.")

  def unlock_doors(self, code):
    if code == "123456":
       print("Doors unlocked")
    else:
      print("Incorrect code. Doors remain locked")

class Bike(Vehicle):
  def __init__ (self, make,year,mileage, bike_type):
    super().__init__(make,year,mileage)
    self.bike_type=bike_type

  def display_info(self):
    super().display_info()
    print(f"Bike Type: {self.bike_type}")

  def pedal(self, effort_level):
    if effort_level >=5:
      print("You are pedaling hard!")
    else:
      print("Steady pedaling. Enjoy the ride")

```

```python
car=Car("Toyota", 2025, 11000, 4, "Gasoline")
car.display_info()
car.drive(100)
car.unlock_doors("123456")
```

```python
bike=Bike("Trek", 2021, 500, "Mountain Bike")
bike.display_info()
bike.pedal(6)
```

```python
bike.drive(10)
```
