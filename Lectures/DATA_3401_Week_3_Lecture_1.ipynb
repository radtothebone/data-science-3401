Question 1: Checking the leap Year
A year is a leap year if the year is disible by 4 and not divisible by 100, or the year is divisible by 400.

ex: 1900 is not a leap year. 2000 is a leap year. 2024 is a leap year.

```python
year = 2024

if(year%4 ==0 and year%100 !=0) or (year%400 == 0):
   print("This year is a leap year")
else:
   print("This year is not a leap year")
```

Question 2: Create an basic calculator

```python
num_1= float(input("Enter the first number:"))
op = input("Enter Operatior:")
num_2= float(input("Enter the second number:"))

if op=="+":
  print(num_1+num_2)
elif op=="-":
  print(num_1-num_2)
elif op=="*":
  print(num_1*num_2)
elif op=="/":
  print(num_1/num_2)
elif op=="**":
  print(num_1**num_2)
else:
  print("Invalid Operation")
```

Nested If Statement

```python
# checking your age groups:

age=69

if age<18: # age<18
   if age<13:  # age<13<18
      print("Child")
   else: # 13<age<18
      print("Teenager")
else:
   if age<65: # 18<age<65
      print("Adult")
   else: # age>65
      print("Senior")
```
