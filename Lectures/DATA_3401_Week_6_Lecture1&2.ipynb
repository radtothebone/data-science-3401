```python
# Class Definition
class Pizza:
  def __init__(self, name):
    self.name=name # Name of the pizza
    self.toppings=[] #list to store toppings

  #medthod to add toppings
  def add_topping(self, topping):
    self.toppings.append(topping)

  #method to display pizza details
  def display(self):
    return f"{self.name} pizza with topping: {', '.join(self.toppings)}"
```

```python
# Margherita
margherita = Pizza("Margherita")
margherita.add_topping("bell pepper")
margherita.add_topping("mushroom")
print(margherita.display())

```

```python
# Pepperoni
pepperoni=Pizza("Pepperoni")
pepperoni.add_topping("Cheese")
pepperoni.add_topping("Pepperoni")
print(pepperoni.display())
```

```python
class Car:
  def __init__(self, color, brand, speed):
    self.color=color
    self.brand=brand
    self.speed=speed
    self.fuel_level=100

  def drive(self):# simulates driving the car. When called, it decreases the fuel level.
    self.fuel_level -=10 #Decreases the fuel level by 10 units
    print("The car is driving.") #Prints message indicating the car is driving.

  def stop(self): #simulates stopping the car.
    print("The car is stopped.")

  def refuel(self):
    self.fuel_level=100 # resets the fuel level to 100
    print("Car is refueled.")




```

```python
my_car=Car("red",'Toyota',60)
my_car.drive()
my_car.stop()
my_car.refuel()
```

```python
class Smartphone:
  def __init__(self, brand,model,color,battery_level,storage_capacity):
    self.band=brand
    self.model=model
    self.color=color
    self.battery_level=battery_level #Percentage
    self.storage_capacity=storage_capacity #GB

  def make_call(self,number):#  It decreases the battery level by 1%.
    if self.battery_level>0:
        self.battery_level -=1 # Simulate battery usage
        print(f"Calling {number}...")
    else: # If the battery is 0, it prints message asking to charge the phone
        print("Battery is too low, Please charge the phone.")

  def send_message(self,number, message): #It decreases the battery level by 1%.
    if self.battery_level>0:
        self.battery_level -=1 # Simulate battery usage
        print(f"Sending meaasge to {number}:{message}")
    else: # If the battery is 0, it prints message asking to charge the phone
        print("Battery is too low, Please charge the phone.")

  def take_photo(self):
     if self.battery_level >0:
         self.battery_level -=5
         print("Photo taken")
     else:
         print("Battery is too low, Please charge the phone.")

  def charge(self):
    self.battery_level=100
    print("Phone is fully charged")


```

```python
import math

class ListMath:
  def __init__(self,numbers):
    self.numbers=numbers

  def display(self):
    print("Current List:", self.numbers)

  def max_value(self):
    return  max(self.numbers)

  def average(self):
    return sum(self.numbers)/len(self.numbers)

  def standard_deviation(self):
    mean=self.average()
    variance=sum((x-mean)**2 for x in self.numbers)/len(self.numbers)
    return math.sqrt(variance)

  def add_number(self, number):
    if isinstance(number, (int, float)):
      self.numbers.append(number)
      print(f"Added{number}.")
    else:
      print("Only integers or floats can be added.")
```

```python
data=ListMath([1,4,-2,9,-5,16])
```

```python
data.display()
```

```python
data.max_value()
```

```python
data.average()
```

```python
data.standard_deviation()
```

```python
data.add_number("Hello")
```
