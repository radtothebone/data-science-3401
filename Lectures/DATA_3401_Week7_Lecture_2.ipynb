# Example: A Universitiy system

```python
# Base Class: Person

class Person:
   def __init__(self, name, age):
     self.name = name
     self.age = age

   def display_info(self):
     print(f"Name: {self.name}, Age: {self.age}")

   def celebrate_birthday(self):
     self.age += 1
     print(f"{self.name} has celebrated their {self.age}th birthday!")
```

```python
# Subclass: Student
class Student(Person):
   def __init__(self, name, age, student_id):
    super().__init__(name,age)
    self.student_id=student_id
    self.courses=[] #empty list for courses

   def enroll(self, course):
    if course not in self.courses:
       self.courses.append(course)
       print(f"{self.name} has enrolled in {course}")
    else:
       print(f"{self.name} is already enrolled in {course}")

   def display_courses(self):
     if self.courses:
       print(f"{self.name}'s Courses:")
       for course in self.courses:
         print(f"-{course}")
     else:
       print(f"{self.name} is not enrolled in any courses.")
```

```python
# Subclass: Instructor
class Instructor(Person):
  def __init__(self, name, age, employee_id):
    super().__init__(name,age)
    self.employee_id=employee_id
    self.courses_taught= []

  def assign_teaching(self, course):
     if course not in self.courses_taught:
       self.courses_taught.append(course)
       print(f"{self.name} is now teaching {course}.")
     else:
       print(f"{self.name} is already teaching {course}.")

  def display_courses_taught(self):
     print(f"{self.name}'s Courses Taught:")
     for course in self.courses_taught:
       print(f"-{course}")
```

```python
class University:
  def __init__(self, name):
    self.name=name
    self.students=[]
    self.instructors=[]

  def add_student(self,student):
    self.students.append(student)

  def add_instructor(self,instructor):
    self.instructors.append(instructor)

  def find_person_by_name(self,name):
     for person in self.students+self.instructors:
       if person.name==name:
         return person
     print(f"No person found with the name{name}")

  def display_all_people(self):
     print(f"All people at {self.name}:")
     for person in self.students+self.instructors:
        person.display_info()
```

```python
Alice=Student("Alice", 20, "DATA 3401")
Brian=Instructor("Brian", 39, "DATA 3442")

Alice.enroll("DATA 1301")
Alice.enroll("DATA 3311")

Brian.assign_teaching("DATA 3402")
Brian.assign_teaching("ASDS 6304")
```

```python
uni=University("My University")
uni.add_student(Alice)
uni.add_instructor(Brian)
uni.display_all_people()
```

# Example: Tic-Tac-Toe Game

```python
class Player: # The Player class represents a player in the Tic-Tac-Toe game.
  def __init__(self,symbol):
    self.symbol=symbol

  def make_move(self, board, row, col):
      # First, it checks if the selected space on the board is free.
      # 'board' is an instance of the Board class, and row and col are the move coordinates.
    if board.is_space_free(row,col):
      board.place_symbol(row, col, self.symbol)
      # After placing the symbol, return True indicating the move was successful.
      return True
    return False

class Board: # The Board class represents the 3x3 Tic-Tac-Toe board.
  def __init__(self):
   self.board=[] # An empty list to store the rows.
   for i in range(3): # Loop 3 times to create 3 rows.
     row=[' ',' ',' '] # Create a row with 3 empty spaces.
     self.board.append(row) # Add the row to the board.

  def display(self): # This loop iterates through each row of the board.
    for row in self.board:
      print("|".join(row))# It prints the symbols in the row, separating them with "|", which simulates the grid lines.
      print("-"*5) #It then prints a row separator line ("-") to visually separate rows in the grid.

  def is_space_free(self, row,col):
    return self.board[row][col]==' '#Here is a space, not nothing. If the space is empty, the method returns True.

  def place_symbol(self,row,col,symbol):
    self.board[row][col]=symbol # This method sets the cell at position (row, col) to the player's symbol.

  def check_winner(self,symbol):
    for i in range(3):
       if self.board[i][0]==symbol and self.board[i][1]==symbol and self.board[i][2]==symbol:
         return True # If all symbols in a row are the same, the player wins.
       if self.board[0][i]==symbol and self.board[1][i]==symbol  and self.board[2][i]==symbol:
         return True # If all symbols in a column are the same, the player wins.

    if self.board[0][0]==symbol and self.board[1][1]==symbol and self.board[2][2]==symbol:
      return True # The player wins if all symbols on the main diagonal match.

    if self.board[0][2]==symbol and self.board[1][1]==symbol and self.board[2][0]==symbol:
      return True # The player wins if all symbols on the anti-diagonal match.
    return False # If no winning condition is met, return False. indicating that no player has won yet.
```

```python
def play_game():
  board=Board()
  player1=Player("X")
  player2=Player("O")
  current_player=player1

  while True:
    board.display()
    #The current player (either 'X' or 'O') is prompted to enter the row and column
    row= int(input(f"{current_player.symbol}'s turn. Enter row:"))
    col= int(input("Enter column:"))
    if not current_player.make_move(board,row,col):
      print("Invalid move. Try again.")
      continue
#The continue statement is used to skip the rest of the loop and ask the player for a new move.
#After a valid move, the game checks if the current player has won by calling the check_winner() method on the Board object.
#and the break statement exits the loop, ending the game.
    if board.check_winner(current_player.symbol):
      print(f"{current_player.symbol} wins!!")
      break
#If the board is full and no winner is found, it's declared a draw
    if all(all(cell !=' ' for cell in row) for row in board.board):
      print("It is a draw!")
      break

    current_player=player2 if current_player==player1 else player1

  board.display()#Displays the final state of the board:
```

```python
# To play the game
play_game()
```
