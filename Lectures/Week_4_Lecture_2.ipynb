###Counting Characters
 Write a Python program that asks the user to enter a sentence. The program should count and display the total number of characters (including spaces) in the sentence.

```python
def count_characters():
  sentence=input("Enter a sentence:")
  count=len(sentence)
  print("Total characters:", count)

count_characters()
```

### Counting Numbers

Write a Python program that prompts the user to enter a series of numbers.

The program should keep accepting numbers until the user enters the keyword 'done'. After that, it should display the total count of numbers entered.

```python
def count_num():
  count=0
  num= input("Enter a number(or 'done' to finish):")

  while num !='done':
    count+=1 # count=count+1
    num= input("Enter a number(or 'done' to finish):")

  print("Total numbers entered:", count)

count_num()
```

## **Passing an Arbitrary Number of positional Arguments**

**Sometimes you won’t know ahead of time how many arguments a function needs to accept.**


- These arguments are captured into a tuple.


**For example, consider a function that builds a pizza**.


- It needs to accept a number of toppings, but you can’t know ahead of time how many toppings a person will want.

## Tuple

When the number of arguments are unknown for a function, They can all be packed into a tuple as shown:

```python
### Example 1:(basic idea of tuple)

def example_func(*args):
   for arg in args:
     print(arg)

example_func(1,2,3,4,5,6)
```

```python
def printALL(*args):
  print("No. of arguments,",len(args))

  for i in args:
    print(i)
printALL("Horse","cat","dog","Monkey") # printALL with 4 arguments
printALL("Arlington","Dallas","Fort Worth") # printALL with 3 arguments
```

```python
def printDictionary(**args):# two star for dictionary Tuple
  for key in args:
    print(key+ ":"+ args[key])# Dictionary

printDictionary(Country="America",State='Texas', City='Arlington')
# Country, State and City are keys. "America", "Texas",and "Arlington" are args.
```

```python
## Example: Building a sandwish

def build_sandwish(size,*fillings):
  print("Making a "+ size + "Sandwish with following fillings:")
  for filling in fillings:
     print("- "+ filling)

build_sandwish("Large ", "mortadella","pesto" )
```

```python
def build_pizza(size,*toppings):
  print("Making a "+ str(size) + "-inch pizza with following toppings:") # convert size as a string.
  for topping in toppings:
     print("- "+ topping)

build_pizza(12, "pepperoni" )
build_pizza(14, "mushrooms","green peppers","extra cheese")
```

```python
def personal_details(**details):
   for key, value in details.items():
     print(f"{key.capitalize()}: {value}") # .capitalize() Capitalize the key words.

personal_details(name="John",age=32,city="Arlington")
```

```python
### def grade

def record_grade(name, **grades):
   print(f"Grades for {name}:")
   for course, grade in grades.items():
      print(f"{course.capitalize()}:{grade}")

record_grade("Alice",mathematics="A",physics="B",chemistry="A-",histroy="B+")
record_grade("Bob",mathematics="B",physics="C",chemistry="B",histroy="C")
```

## Nested Functions

```python
def Outer_function():
  print("Outer function")

  def inner_function():
    print("Inner function")
  inner_function()

Outer_function()
```

```python
def rectangle(l,w):
  def area():
    return l*w
  def perimeter():
    return 2*(l+w)
  return(area(), perimeter())
```

```python
a, p=rectangle(5,3)
print(a, p)
```
