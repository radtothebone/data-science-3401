Pandas uses two types data structures: Series and DataFrame.

Series: It is a one-dimensional array-like object that can hold any data type.

DataFrame: It is a two dimensional table of data with rows and columns.

1. Using a Numpy Array

```python
import pandas as pd
import numpy as np
data = np.array([
    ['Alice', 25,'New York'],
    ['Brian', 39,'Arlington'],
    ['Bob', 35, 'Dallas']
])
df= pd.DataFrame(data, columns=['Name','Age','Location'])
df
```

2. Using a Dictionary


```python
data={
    'Name':['Alice','Brian','Bob'],
    'Age':[25,39,35],
    'Location':['New York','Arlington','Dallas'],
    'Salary':[50000, 60000, 70000]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df[['Name','Age','Salary']])
```

## loc and iloc
1. loc: based on the row and column labels
   data.loc[row_label, column_label].
   (inclusive both start and end values in a range)

2. iloc:based on the integer index positions.
   data.iloc[row_index,column_index]
   (Does not include the last value in a range, similar to typical Python slicing.)

```python
print(df.loc[1:2])
```

```python
print(df.iloc[1:2])
```

```python
print(df.loc[1:2,['Name','Location']])
```

```python
print(df.iloc[1:2,0:3])
```

```python
print(df.loc[df['Age']>30])
```

```python
df.loc[2,'Location']='Fort Worth'# selection is based on the name.
print(df)
```

```python
df.iloc[1,2]='Arlington'# selection is based on the index.
print(df)
```

```python
print("Summary Statistics:")
print(df.describe())
```

```python
print(df['Salary'].mean())
```

```python
data={
    'Name':['Alice','Brian','Bob','Jordan','Eric','David'],
    'Age':[25,39,35,28,32,30],
    'Location':['New York','Arlington','Dallas','Tokyo',' LA','Sydney'],
    'Salary':[50000, 60000, 70000, 55000,65000,55000]
}
df=pd.DataFrame(data)
print(df)
```

```python
filtered_df=df[df['Salary']>55000]
print(filtered_df)
```

```python
sorted_df=df.sort_values('Salary')
print(sorted_df)
```

```python
data2={'A':[10,20,30]}
df2=pd.DataFrame(data2)
df2.index +=1 #shift index to start from 1
print(df2)
```

```python
# Finding the number of Missing values
print(df2.isnull())
```

```python
# Find the number of missing values in each column
missing_values=df.isnull().sum()
print(missing_values)
```

```python
#Calculate the total number of missing values for the entire dataset
total_missing_values=missing_values.sum()
print(total_missing_values)
```

```python
# Drop the rows with missing values
df_dropped=df.dropna()
print(df_dropped)
```

```python
# Drop the missing value by Columns
df_dropped_columns=df.dropna(axis=1)
print(df_dropped_columns)
```
