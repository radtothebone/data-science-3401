Boolean Data Type

```python
x=10
y=15
greater = x>=y
less_equal= x<=y
logical_and = greater and less_equal
logical_or = greater or less_equal
logical_not = not greater
print(greater)
print(less_equal)
print(logical_and)
print(logical_or)
```

```python
x=10
y=15
equal= x==y
not_equal= x!=y
greater_than = x>y
less_than_equal = x<=y
print(equal)
print(not_equal)
print(greater_than)
print(less_than_equal)
```
