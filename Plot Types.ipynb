```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9,7))

# Pie chart
sizes = (30, 20, 20, 20, 30)
labels = ('Swag', 'Dingus', 'Bro', 'XXXTENTACION', 'TWEAKER')
colors = ('red', 'blue', 'green', 'orange', 'purple')
explode = (0, 0.2, 0, 0, 0)

plt.pie(sizes, labels = labels, colors = colors, explode=explode, autopct='%1.2f%%', startangle=30)
plt.show()
```

```python
# Hexbin Plot
import numpy as np

x = np.random.rand(10000)
y = np.random.rand(10000)

plt.figure(figsize=(9,6))
plt.hexbin(x, y, gridsize=10, cmap='plasma', bins=9)
plt.colorbar()
plt.show()
```

```python
# Subplots
x = np.linspace(0, 2*np.pi, 100)

y1 = np.sin(x)
y2 = np.cos(x)

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title('subplots')
plt.ylabel('Sin(x)')

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title('Subplot 2')
plt.ylabel('Cos(x)')

plt.tight_layout()
```

```python
# Row example

x = np.linspace(0, 5, 10)
y = x**2

plt.subplot(1, 2, 1)
plt.plot(x, y, 'r*-')

plt.subplot(1, 2, 2)
plt.plot(y, x, 'g*-')
```
