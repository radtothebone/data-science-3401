```python
import pandas as pd
import numpy as np

data = {
    'Name':['Alice', 'Brian', 'Bob'],
    'Age': [25, 39, 35],
    'Location': ['New York', 'Arlington', 'Dallas'],
    'Salary': [50, 20, 500]
}

df = pd.DataFrame(data)

display(df)

# Finding the shape of the dataframe.
print('shape:', df.shape, '\n')

# Showing column names, note the first 
# value in the dict pair is what the column name is.
print(df.columns, '\n')

# Printing values of a list of columns
print(df[['Name', 'Age']], '\n')

# loc, based on row and column labels.
print(df.loc[0,'Name'])

# iloc, based on integer indecies.
print(df.iloc[:,:])
```
