```python
# The super class
class Pizza:
    def __init__(self, name):
        self.name = name
        self.topping = []
        
    def add_topping(self, topping):
        self.topping.append(topping)
        
    def bake(self):
        print(f'Baking {self.name} Pizza with topping, {self.topping}')
```

```python
# Defines another subclass
class MargheritaPizza(Pizza):
    def __init__(self):
        super().__init__('Margherita') # Calling the parent initializer
        self.add_topping('Tomato')
        self.add_topping('Mozzarella')
```

```python
# Defining another subclass
class PepperoniPizza(Pizza):
    def __init__(self):
        super().__init__('Pepperoni') # Parent initializer
        self.add_topping('Pepperoni')
        self.add_topping('Tomato')
        self.add_topping('Cheese')
```

```python
pepperoni = PepperoniPizza()
```

```python
class Vehicle: 
    def __init__(self, make, year, mileage):
        self.make = make
        self.year = year 
        self.mileage = mileage

    def display_info(self):
        print(f'{self.make} {self.year} {self.mileage}')
        
    def start_engine(self):
        print('The engine has started.')
        
    def set_mileage(self, mileage):
        if mileage >= self.mileage:
            self.mileage = mileage
            print(f'New mileage set to {self.mileage} miles.')
        else:
            print('Error, new mileage cannot be less than current mileage.')
            
    def get_mileage(self):
        return self.mileage
        
    def drive (self, distance):
        if distance > 0:
            self.mileage += distance
            print(f'Driving {distance} miles. New mileage is {self.mileage} miles.')
        else:
            print('Distance must be greater than zero.')
```

```python
# Car subclass
class Car(Vehicle):
    def __init__(self, make, year, mileage, doors, fuel_type):
        super().__init__(make, year, mileage)
        self.doors = doors
        self.fuel_type = fuel_type
        
    def display_info(self):
        super().display_info()
        print(f'{self.doors}, {self.fuel_type}')
        
    def lock_doors(self):
        print('Locked the doors.')
        
    def unlock_doors(self, code):
        if code == '12345':
            print('doors unlocked')
        else:
            print('Incorrect code, initializing self destruct.')
```
