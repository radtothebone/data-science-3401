## **Space Adventure** (100 Points)


### **Objective:**
You have to extend the space mission by adding different types of spaceships and astronauts. The goal is to use **inheritance** to create specialized subclasses for a more advanced space mission. You'll implement classes with attributes and methods, then override them in subclasses to customize the behavior of each type of spaceship and astronaut.

#### Tasks:

1. **Spaceship Class (Superclass) (20 Points)**:
   - Create a superclass named `Spaceship` with the following attributes:
     - `name` (string): The name of the spaceship.
     - `fuel` (integer): The spaceship’s starting fuel, initialized to 100.
   - Implement the following methods:
     - `display_status()`: Prints the spaceship’s name and fuel level.
     - `travel(distance)`: Reduces fuel based on the `distance` traveled. For every 10 units of distance, reduce fuel by 5 units. If fuel reaches zero, print that the spaceship has run out of fuel.
     - `refuel()`: Refuels the spaceship by 20 units, but the fuel cannot exceed 100.

2. **CargoSpaceship Class (Subclass of Spaceship) (20 Points)**:
   - Create a subclass named `CargoSpaceship` that inherits from `Spaceship` with an additional attribute:
     - `cargo_capacity` (integer): The maximum amount of cargo the spaceship can carry.
   - Override the `display_status()` method to also print the cargo capacity.
   - Add a new method `load_cargo(amount)` to load cargo. If the cargo exceeds the capacity, print a warning message.

3. **Astronaut Class (Superclass) (20 Points)**:
   - Create a superclass named `Astronaut` with the following attributes:
     - `name` (string): The name of the astronaut.
     - `oxygen_level` (integer): The astronaut’s oxygen level, initialized to 100.
   - Implement the following methods:
     - `display_info()`: Prints the astronaut’s name and oxygen level.
     - `perform_spacewalk()`: Decreases oxygen level by 20. If oxygen falls below 20, print a warning that the astronaut is running out of oxygen.
     - `replenish_oxygen()`: Increases oxygen by 20 units, but it cannot exceed 100.

4. **EngineerAstronaut Class (Subclass of Astronaut) (20 Points)**:
   - Create a subclass named `EngineerAstronaut` that inherits from `Astronaut` with an additional attribute:
     - `repair_skill` (integer): The astronaut’s ability to repair the spaceship, initialized to 50.
   - Override the `perform_spacewalk()` method to also print the astronaut’s repair skill after the spacewalk.
   - Add a new method `repair(spaceship)` to repair the spaceship’s fuel by 10 units, but fuel cannot exceed 100.

5. **Mission Simulation (20 Points)**:
   - Implement a `space_mission()` function that:
     - Creates one `CargoSpaceship` and one `EngineerAstronaut` object.
     - The spaceship must travel a total distance of 50 units.
     - At the midpoint (25 units), the spaceship will need to refuel, and the astronaut will perform a spacewalk and repair the spaceship.
     - The mission ends when the spaceship completes the 50-unit journey or if the spaceship runs out of fuel or the astronaut runs out of oxygen.






### **Example Output:**

```python
Spaceship 'Cargo Carrier' is ready for the mission!
Cargo Capacity: 500

Traveling to the distant planet...

Traveling 25 units...
Fuel remaining: 87

Astronaut Alice performs a spacewalk. Oxygen level: 80, Repair skill: 50

Refueling the spaceship...
Fuel is now 100

Repairing the spaceship... Fuel increased by 10 units. Fuel is now 100

Traveling another 25 units...
Fuel remaining: 87

Mission Complete! You have reached the planet!
```

### **Solution:**

```python
class Spaceship:
    def __init__(self, name, fuel=100):
        self.name = name
        self.fuel = fuel
        
    def display_status(self):
        print(f'Spaceship name: {self.name}')
        print(f'Fuel Level: {self.fuel}')
        
    def travel(self, distance):
        self.fuel -= distance * 0.5
        if self.fuel <= 0:
            self.fuel = 0
            print('You ran out of gas. ')
            
    def refuel(self):
        self.fuel += 20
        if self.fuel > 100:
            self.fuel = 100
            
            
class CargoSpaceship(Spaceship):
    def __init__(self, name, cargo_capacity, fuel=100):
        super().__init__(name, fuel)
        self.cargo_capacity = cargo_capacity
        self.cargo_loaded = 0
        
    def display_status(self):
        print(f'Spaceship name: {self.name}')
        print(f'Fuel Level: {self.fuel}')
        print(f'Cargo Capacity: {self.cargo_capacity}')
        print(f'Cargo Loaded: {self.cargo_loaded}')
        
    def load_cargo(self, amount):
        self.cargo_loaded += amount
        if self.cargo_loaded > self.cargo_capacity:
            print('Warning, you are overweight. ')


class Astronaut:
    def __init__(self, name, oxygen_level=100):
        self.name = name
        self.oxygen_level = oxygen_level

    def display_info(self):
        print(f'Astronaut name: {self.name}')
        print(f'Oxygen Level: {self.oxygen_level}')

    def perform_spacewalk(self):
        self.oxygen_level -= 20
        if self.oxygen_level < 0:
            self.oxygen_level = 0
        if self.oxygen_level < 20:
            print("The astronaut\'s oxygen level is low. '")
    def replenish_oxygen(self):
        self.oxygen_level += 20
        if self.oxygen_level > 100:
            self.oxygen_level = 100


class EngineerAstronaut(Astronaut):
    def __init__(self, name, oxygen_level=100, repair_skill=50):
        super().__init__(name, oxygen_level)
        self.repair_skill = repair_skill

    def perform_spacewalk(self):
        self.oxygen_level -= 20
        if self.oxygen_level < 0:
            self.oxygen_level = 0
        if self.oxygen_level < 20:
            print("The engineer astronaut\'s oxygen level is low. '")
        print(f'Repair Skill: {self.repair_skill}')

    def repair(self, spaceship):
        spaceship.fuel += 10
        if spaceship.fuel > 100:
            spaceship.fuel = 100

        print('Engineer has repaired the spaceship. +10 fuel.')


def space_mission():
    ship = CargoSpaceship('Atlas', cargo_capacity=100)
    astro = EngineerAstronaut('Christian')
    total_distance = 50
    traveled = 0
    print('Start mission!')
    ship.display_status()
    astro.display_info()

    while traveled < total_distance:
        step = 25 if traveled == 0 else 25
        print(f'Traveling {step} units. ')
        ship.travel(step)
        traveled += step
        ship.display_status()

        if ship.fuel <= 0:
            print('Mission failed: spaceship ran out of fuel.')
            break

        if traveled == 25:
            print('Midpoint section')
            print('Refueling spaceship.')
            ship.refuel()
            print('Astronaut performing spacewalk.')
            astro.perform_spacewalk()

            if astro.oxygen_level <= 0:
                print('Mission failed: astronaut ran out of oxygen.')
                break

            print('Astronaut repairing spaceship.')
            astro.repair(ship)
            ship.display_status()
            astro.display_info()

    if traveled >= total_distance and ship.fuel > 0 and astro.oxygen_level > 0:
        print('\nMission success: spaceship completed the 50-unit journey!')

    print('Mission Complete.')


space_mission()
```
