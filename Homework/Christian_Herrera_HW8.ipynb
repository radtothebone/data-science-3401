### **Question 11:** (15 Points)

Create a function that asks the user for a numeric input to specify how many random integers to generate in a 1D array (between 1 and 100). Identify all prime numbers in the array and replace them with the number 1.

````python
def is_prime(n):
    # Prime? 
    if n < 2:
        return False
    
    for i in range(2, n):
        if n % i == 0:
            return False
    
    return True

def replace_primes_with_one():
    # Replace primes with one. 
    size = int(input("How many random integers do you want? "))
    
    arr = np.random.randint(1, 101, size)
    
    print("Original array:")
    print(arr)
    
    # All of them. 
    for i in range(len(arr)):
        if is_prime(arr[i]):
            arr[i] = 1
    
    # Here you go. 
    print("Updated array:")
    print(arr)

replace_primes_with_one()
````

# **Question 12: Number Guessing Game** (25 Points)

### Objective
The objective of this homework assignment is to create a Number Guessing game using Python. This will help you practice and reinforce your understanding of basic programming concepts such as conditionals, loops, functions, user input, and random number generation.

### Instructions

#### 1. Setup the Project
- Create a new Python file named `number_guessing_game.py`.

#### 2. Game Requirements
- The game should randomly select a number within a specified range.
- The user should be prompted to guess the number.
- The game should provide feedback if the guess is too high, too low, or correct.
- The game should continue until the user guesses the correct number or decides to stop playing.

#### 3. Detailed Steps to Implement

1. **Import the Necessary Module**
   - Use the `random` module to generate a random number.
   ```python
   import random

2. **Generate the Random Number**

- Define a function generate_random_number(min_value, max_value) to generate a random number within the given range.

3. **Get User Guess**

- Define a function get_user_guess() to prompt the user to enter their guess and validate the input.

4. **Check Guess**

- Define a function check_guess(guess, target) to check if the guess is too high, too low, or correct.

5. **Main Game Function**

- Define the main game function play_game() to implement the game loop and manage the game logic.

````python
# Grab a random number.
def generate_random_number(min_value, max_value):
    return random.randint(min_value, max_value)

# Ask the user for a guess.
def get_user_guess():
    while True:
        guess = input("Enter your guess: ")
        
        # Make sure it is actually a number.
        if guess.isdigit():
            return int(guess)
        else:
            print("Please enter a valid number.")

# See how bad the guess was.
def check_guess(guess, target):
    if guess < target:
        return "too low"
    elif guess > target:
        return "too high"
    else:
        return "correct"

# Main event.
def play_game():
    min_value = 1
    max_value = 100
    
    # Secret number. Very classified.
    target = generate_random_number(min_value, max_value)
    
    print("Welcome to the Number Guessing Game!")
    print(f"I'm thinking of a number between {min_value} and {max_value}.")
    
    while True:
        # Get the user's heroic attempt.
        guess = get_user_guess()
        
        # Check the damage.
        result = check_guess(guess, target)
        
        if result == "too low":
            print("Too low. Try again.")
        elif result == "too high":
            print("Too high. Try again.")
        else:
            print("Correct! You guessed the number.")
            break
        
        # In case they are tired of losing.
        stop = input("Do you want to keep playing? (yes/no): ").lower()
        
        if stop == "no":
            print("Game stopped.")
            break

# Start the chaos.
play_game()
````
