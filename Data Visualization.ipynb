```python
import matplotlib.pyplot as plt
import pandas as pd 

data = { 'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022], 
        'Sales': [1000, 1500, 1200, 1800, 2000, 900, 800, 2200], 
        'Category': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B'], 
        'Revenue': [50000, 60000, 70000, 55000, 65000, 35000, 40000, 90000], 
        'Age': [25, 30, 35, 28, 32, 34, 39, 43], 
        'Salary': [50000, 60000, 70000, 55000, 65000, 78000, 62000, 80000], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Male', 'Female', 'Male'] } 

df = pd.DataFrame(data)
```

```python
from matplotlib.lines import lineStyles
plt.plot(df['Year'], df['Sales'], linestyle = '--')
```

```python
# More complex plot

# Figure with specific size
# Starting with figure is like starting with a canvas. 
plt.figure(figsize=(20,15))

plt.plot(df['Year'], df['Sales'], 
         linestyle = '--', 
         linewidth=4, 
         color='teal', 
         marker='*', 
         markersize=20)

plt.grid(True, linewidth=2)

plt.xticks(df['Year'])
plt.yticks(range(700, 2300, 200))

plt.show()

plt.show()
```
